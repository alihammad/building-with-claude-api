```python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)
```

```python
def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)
```

```python
import textwrap

# add in the initial user question
messages = []
add_user_message(messages, "Generate a one paragraph killer AI product idea for Equifax and other credit rating agencies.")

with client.messages.stream(
    model=model,
    max_tokens=1000,
    messages=messages,
) as stream:
    for text in stream.text_stream:
        print(textwrap.fill(text, width=80), end="")
        pass

# final = stream.get_final_message()
# content = ""
# if isinstance(final, dict):
#     content = final.get("content") or final.get("text") or str(final)
# else:
#     content = str(final)

# print("\n\nFinal message:")
# print(textwrap.fill(content, width=80))
# stream.get_final_message()
```
