```python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-0"
```

```python
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "Write a short poem about the beauty of nature."            
        }
    ],
)
```
